```python
a=float(input("Введите первое число"))
op=input("Введите операцию (+,-,*,/)")
b=float(input("Введите второе число"))
if op == '+':
    r = a + b
elif op == '-':
    r = a - b
elif op == '*':
    r = a*b
elif op == '/':
    if b != 0:
        r = a / b
    else:
        r = "Ошибка! (на ноль делить нельзя!)"
else:
    r = "Невозможная операция"
print("Результат:", r)

```

```python
x=int(input("Введите сумму вклада (в рублях):"))
p=int(input("Введите процент вклада:"))
y=int(input("Введите целевую сумму (в рублях):"))
m=x*100
sum=y*100
years=0
while m<sum:
    m += (m*p)//100
    years += 1
print("Количество лет:", years)


```

```python
m1=0
m2=0
while True:
    n=int(input("Введите число (0-конец последовательности)"))
    if n == 0:
        break
    if n > m1:
        m2=m1
        m1=n
    elif n > m2:
        m2 = n
print("2-e по величине число:", m2)
```
