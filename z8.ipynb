```python
string = ("Hello", "Holla")
set1 = set(string[0])
set2 = set(string[1])
sim = set1.intersection(set2)
diff = set1.difference(set2)
print("Общие буквы:", sim)
print("Разные буквы:", diff)

```

```python
s = []
while True:
    num=int(input("Введите число:"))
    if num == 0:
        break
    s.append(num)
for n in s:
    if n < 0:
        s = [abs(x) for x in s]
        break
print(s)
```

```python
scrable = {
    1: "AEIOULNSTRАВЕИНОРСТ",
    2: "DGДКЛМПУ",
    3: "BCMPБГЁЬЯ",
    4: "FHVWYЙЫ",
    5: "KЖЗХЦЧ",
    8: "JXШЭЮ",
    10: "QZФЩЪ"
}
word = input("Введите слово: ").upper()
sum = 0
for letter in word:
    for points, letters in scrable.items():
        if letter in letters:
            sum += points
            break
print("Стоимость:", sum)
```

```python
input_list = [1,2,3,4,5,6]
sum_chet_pos = 0
sum_nechet_pos = 0
for i, value in enumerate(input_list):
    if i % 2 == 0:
        sum_chet_pos += value
    else:
        sum_nechet_pos += value       
print(sum_chet_pos / sum_nechet_pos)
```

```python
input_list = ["kar", "meow", "https://rsmu.ru", "bioinf", "http://yandex.ru", "Russia"]
print(f"Исходный список {input_list}")
result_list = input_list.copy()
for item in input_list:
    if not item.startswith("http://"):
        result_list.remove(item)
print(f"Осталось {result_list}")
```

```python
import random
import time
N = 10
num = []
for k in range(N):
    random_number = random.randint(1, 10)
    num.append(random_number)
print("Исходный массив: ", num)
for i in range(N):
    for j in range(N-1-i):
        if num[j] > num [j + 1]:
            num[j], num[j + 1] = num[j + 1], num[j]
print("Отсортированный массив: ", num)

lenght = [10, 100, 1000, 10000]
for size in lenght:
    test = []
    for k in range(size):
        test.append(random.randint(1, 10))
    tm_start = time.time()
    for i in range(size):
        for j in range(size - 1 - i):
            if test[j] > test [j + 1]:
                test[j], test [j + 1] = test[j + 1], test[j]
    tm_end = time.time()
    tm = tm_end - tm_start
    print("Размер списка:", size)
    print("Время работы:", tm)
```

```python
patients = [("Муся", 7,"Игорь", "Бероев"), ("Изольда", 2, "Игорь", "Бероев")]
owners_cats = {}
for name in patients:
    owner = name [2] + " " + name [3]
    cat = name [0] + ", " + str(name [1])
    if owner not in owners_cats:
        owners_cats[owner] = cat
    else:
        owners_cats[owner] = owners_cats[owner] + "; " + cat
for owner in owners_cats:
    print(owner + ": " + owners_cats[owner])
```
